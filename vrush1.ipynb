```python
#regular expression
import re
def validate_pnr(pnr):
  pattern=r'\d{8,10}$'
  if re.fullmatch(pattern,pnr):
    print(f"Valid PNR : {pnr}")
  else:
    print(f"Invalid PNR : {pnr}")
```

```python
validate_pnr("27364857")
validate_pnr("8364A638")     #contains letter
validate_pnr("5278356")      #only 7 characters
validate_pnr("29576034")
```

```python
def validate_seat(seat):
    pattern = r'[A-Z][0-9]{1,2}$'
    if re.fullmatch(pattern, seat):
        print("Valid Seat Number :", seat)
    else:
        print("Invalid Seat Number :", seat)
```

```python
validate_seat("B4")
validate_seat("A15")
validate_seat("b4")     #lowercase letter
validate_seat("234")    #wrong order
validate_seat("A132")   #extra number
```

```python
def extract_train_number(text):
  pattern=r'\d{4,5}'
  results=re.findall(pattern,text)
  print("Train Number Found : ",results)
```

```python
class TicketValidator:
  def __init__(self,pnr,train_number,seat,fare):
    self.pnr=pnr
    self.train_number=train_number
    self.seat=seat
    self.fare=fare

  def validate(self):
    print("="*40)
    print("Indian Railway Ticket Validation Report")
    print("="*40)

    if re.fullmatch(r'[0-9]{8}$',self.pnr):
      print("PNR : ",self.pnr,"-> VALID")
    else:
      print("PNR : ",self.pnr,"-> INVALID")

    if re.fullmatch(r'[0-9]{4,5}$',self.train_number):
      print("Train Number : ",self.train_number,"-> VALID")
    else:
      print("Train Number : ",self.train_number,"-> INVALID")

    if re.fullmatch(r'[A-Z][0-9]{1,2}$',self.seat):
      print("Seat Number : ",self.seat,"-> VALID")
    else:
      print("Seat Number : ",self.seat,"-> INVALID")


    if re.fullmatch(r'\d{3,4}$',self.fare):
      print("Fare : ",self.fare,"-> VALID")
    else:
      print("fare : ",self.fare,"-> INVALID")
```

```python
ticket1 = TicketValidator("43821756", "12951", "B4",  "880")
ticket2 = TicketValidator("6793452",  "1230X", "b12", "XX")

ticket1.validate()
ticket2.validate()
```

```python
try:
    fare_amount = int(ticket1.fare)
    print("Ticket 1 Fare Conversion : SUCCESS")
except ValueError:
    print("Ticket 1 Fare Conversion : FAILED")
finally:
    print("The Fare is Valid")

try:
    fare_amount = int(ticket2.fare)
    print("Ticket 2 Fare Conversion : SUCCESS")
except ValueError:
    print("Ticket 2 Fare Conversion : FAILED")
finally:
    print("The Fare is Valid")
```
